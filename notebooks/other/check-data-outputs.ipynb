```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train = np.load('../../output/data_eagles_vezhen/lstm-ae/X_train.npy')
y_train = np.load('../../output/data_eagles_vezhen/lstm-ae/y_train.npy')
X_test = np.load('../../output/data_eagles_vezhen/lstm-ae/X_test.npy')
y_test = np.load('../../output/data_eagles_vezhen/lstm-ae/y_test.npy')
X_val = np.load('../../output/data_eagles_vezhen/lstm-ae/X_val.npy')
y_val = np.load('../../output/data_eagles_vezhen/lstm-ae/y_val.npy')
```

```python
val_mmsis = np.load(
    '../../output/data_eagles_vezhen2/val-mmsis.npy',
    allow_pickle=True
)
val_days = np.load(
    '../../output/data_eagles_vezhen2/val-days.npy',
    allow_pickle=True
)

group_results = {}
for ix, (mmsi, day) in enumerate(zip(val_mmsis, val_days)):
    
    if (mmsi, day) not in group_results:
        group_results[(mmsi, day)] = [ y_val[ix, :, :] ]
    else:
        group_results[(mmsi, day)].append( y_val[ix, :, :] )
```
